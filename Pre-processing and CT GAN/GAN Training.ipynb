```python
import pandas as pd
import pandapower as pp
import numpy as np
```

```python
data_file = "Final_Training_Data_15k.csv"

# Load the raw dataset
try:
    df = pd.read_csv(data_file)
    print(f"Data loaded successfully. Rows: {len(df)}, Columns: {len(df.columns)}")
except Exception as e:
    print(f"Error loading file: {e}")
```

```python
df.columns
```

```python
df = df.loc[:, (df != 0).any(axis=0)]
```

```python
df.columns
```

```python
df.shape[1]
```

```python
from sklearn.model_selection import train_test_split

# Split your dataset
train_df = df.iloc[:10000]   # first 10k for training
test_df  = df.iloc[10000:]   # rest for testing
```

```python
train_df.shape, test_df.shape
```

```python
train_df.columns
```

```python
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import SingleTableMetadata
```

```python
# Define metadata (schema) from your dataframe
metadata = SingleTableMetadata()
metadata.detect_from_dataframe(data=df)
```

```python
# Initialize synthesizer
synthesizer = CTGANSynthesizer(metadata, epochs=300,verbose=True)
```

```python
# Train
synthesizer.fit(train_df)
```

```python
synthesizer.save(
    filepath='CT_GAN_FOR_AC_FDIA.pkl'
)
```

```python
synthesizer.get_loss_values()
```

```python
synthetic_df = synthesizer.sample(len(test_df))
```

```python
test_df.head()
```

```python
synthetic_df.head()
```

```python
import pandas as pd

# Compare mean and std side by side
stats_df = pd.DataFrame({
    "Real_Mean": test_df.mean(),
    "Synthetic_Mean": synthetic_df.mean(),
    "Real_Std": test_df.std(),
    "Synthetic_Std": synthetic_df.std()
})

print(stats_df)
```

```python
from sdv.evaluation.single_table import run_diagnostic, evaluate_quality
from sdv.evaluation.single_table import get_column_plot
```

```python
# 1. perform basic validity checks
diagnostic = run_diagnostic(test_df, synthetic_df, metadata)

# 2. measure the statistical similarity
quality_report = evaluate_quality(test_df, synthetic_df, metadata)

```
